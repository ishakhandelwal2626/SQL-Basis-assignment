Q9. What is the Difference Between a Primary Key and a Foreign Key?

Answer - Primary Key

1. A column (or set of columns) that uniquely identifies each record in a table.

2. Cannot contain NULL values.

3. Must be unique.

4. Only one primary key is allowed per table (but it can be composite).

Foreign Key
1. A column that links one table to another.

2. It refers to the Primary Key of another table.

3. Can contain duplicate values.

4. Can contain NULL (unless restricted).

5. A table can have multiple foreign keys.

Q10. What Are Constraints in SQL and Why Are They Used?

Answer - Constraints are rules applied to table columns to control the type of data that can be inserted into a table.

They help maintain accuracy, reliability, and integrity of the data in a database.

Constraints are used to:

1. Ensure data integrity

2. Prevent invalid data entry

3. Maintain relationships between tables

4. Enforce business rules

Examples of constraints - primary key, foreign key, unique, not null, etc.
